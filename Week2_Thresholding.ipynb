```python
import cv2
import os
import numpy as np  # 今天需要用到 numpy 来造一个“刷子”

folder_path = r"D:\ycz32\Desktop\111\Learn_ai\image"
img_path = os.path.join(folder_path, "lenna.jpg")

# 1. 读取灰度图
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

if img is None:
    print("❌ 读取失败")
else:
    # 2. 先用 OTSU 二值化（承接昨天的工作）
    ret, thresh_img = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    
    # 3. 定义“结构元素”（Kernel/核）—— 这就是我们操作的“刷子”
    # np.ones((3, 3), np.uint8) 代表一个 3x3 的全 1 矩阵
    kernel = np.ones((3, 3), np.uint8)

    # 4. 腐蚀操作（让白色变瘦，去噪点）
    eroded_img = cv2.erode(thresh_img, kernel, iterations=1)

    # 5. 膨胀操作（让白色变胖，填补黑洞）
    dilated_img = cv2.dilate(thresh_img, kernel, iterations=1)

    # 6. 保存结果
    cv2.imwrite(os.path.join(folder_path, "eroded_result.jpg"), eroded_img)
    cv2.imwrite(os.path.join(folder_path, "dilated_result.jpg"), dilated_img)
    print("✅ 形态学操作完成并保存！")

    # 7. 显示对比
    cv2.imshow("Original Binary", thresh_img)
    cv2.imshow("Eroded (Thinner)", eroded_img)
    cv2.imshow("Dilated (Fatter)", dilated_img)
    
    print("请在弹出的图片窗口按任意键关闭...")
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2
import os
import numpy as np

folder_path = r"D:\ycz32\Desktop\111\Learn_ai\image"
img_path = os.path.join(folder_path, "lenna.jpg")

img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

if img is None:
    print("❌ 读取失败")
else:
    # 1. OTSU 二值化（承接昨天）
    ret, thresh_img = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

    # 2. 定义 5x5 的核
    kernel = np.ones((5, 5), np.uint8)

    # 3. 开运算（去噪点）
    # 参数 cv2.MORPH_OPEN 表示开运算
    opened_img = cv2.morphologyEx(thresh_img, cv2.MORPH_OPEN, kernel)

    # 4. 闭运算（填黑洞）
    # 参数 cv2.MORPH_CLOSE 表示闭运算
    closed_img = cv2.morphologyEx(thresh_img, cv2.MORPH_CLOSE, kernel)

    # 5. 保存
    cv2.imwrite(os.path.join(folder_path, "opened_result.jpg"), opened_img)
    cv2.imwrite(os.path.join(folder_path, "closed_result.jpg"), closed_img)
    print("✅ 开运算与闭运算完成，结果已保存")

    # 6. 显示对比
    cv2.imshow("Original Binary", thresh_img)
    cv2.imshow("Opened (Remove Noise)", opened_img)
    cv2.imshow("Closed (Fill Holes)", closed_img)
    print("请在弹出的图片窗口按任意键关闭...")
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
import cv2
import os
import numpy as np

# ================= 1. 核心处理函数 =================
def process_image(img_path):
    # 1. 直接读灰度图
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        return None
    
    # 2. OTSU 自动阈值二值化
    ret, thresh = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    
    # 3. 定义 5x5 刷子，做开运算去噪
    kernel = np.ones((5, 5), np.uint8)
    opened = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)
    
    # 4. 返回处理好的最终图
    return opened


# ================= 2. 主流程 =================
def main():
    folder_path = r"D:\ycz32\Desktop\111\Learn_ai\image"
    output_folder = r"D:\ycz32\Desktop\111\Learn_ai\final_binary_output"
    
    # 自动建输出文件夹
    os.makedirs(output_folder, exist_ok=True)
    
    count = 0
    for filename in os.listdir(folder_path):
        if filename.lower().endswith(('.jpg', '.png', '.jpeg')):
            full_path = os.path.join(folder_path, filename)
            
            # 调用流水线处理
            processed_img = process_image(full_path)
            
            if processed_img is None:
                print(f"❌ {filename} 处理失败")
                continue
            
            # 保存到新文件夹
            save_path = os.path.join(output_folder, filename)
            success = cv2.imwrite(save_path, processed_img)
            
            if success:
                count += 1
                print(f"✅ {filename} 预处理完成")
            else:
                print(f"❌ {filename} 保存失败")

    print(f"\n🎉 批量处理完成！共成功处理 {count} 张图片。")

# 运行
main()
```
