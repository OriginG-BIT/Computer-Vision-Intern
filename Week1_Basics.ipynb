```python
import cv2
import os

# ================= 1. 定义核心处理函数 =================
# 输入：一张图片的完整路径
# 输出：处理好的图片（旋转90度+缩放）
def process_image(img_path):
    # 读取图片
    img = cv2.imread(img_path)
    if img is None:
        return None  # 读取失败，直接返回空
    
    # 工业预处理动作：先旋转90度，再缩放
    rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
    resized = cv2.resize(rotated, (512, 512))
    
    # 把结果返回给调用者
    return resized


# ================= 2. 定义主流程 =================
def main():
    folder_path = r"D:\ycz32\Desktop\111\Learn_ai\image"
    output_folder = r"D:\ycz32\Desktop\111\Learn_ai\output"
    
    # 检查输出文件夹是否存在，不存在就建一个
    os.makedirs(output_folder, exist_ok=True)
    
    # 遍历文件夹
    for filename in os.listdir(folder_path):
        if filename.lower().endswith(('.jpg', '.png', '.jpeg')):
            full_path = os.path.join(folder_path, filename)
            
            # 调用函数，拿到处理好的图片
            processed_img = process_image(full_path)
            
            # 如果没处理成功，跳过
            if processed_img is None:
                print(f"❌ {filename} 读取或处理失败")
                continue
            
            # 保存到新文件夹
            save_path = os.path.join(output_folder, filename)
            success = cv2.imwrite(save_path, processed_img)
            
            # 打印结果
            print(f"✅ {filename} 处理完成，保存结果：{success}")

# 运行主流程
main()
```

```python
import cv2
import os

# ================= 1. 核心处理函数 =================
# 把这一周学的所有预处理动作，打包成一个函数
def process_image(img_path):
    # 1. 读取图片
    img = cv2.imread(img_path)
    if img is None:
        return None
    
    # 2. 转灰度图（工业视觉常用预处理）
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    # 3. 旋转90度
    rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
    
    # 4. 缩放成 512x512 标准尺寸
    resized = cv2.resize(rotated, (512, 512))
    
    # 5. 提取中心区域的 ROI（比如中心 200x200）
    h, w = resized.shape[:2]
    center_y, center_x = h // 2, w // 2
    roi = resized[center_y-100:center_y+100, center_x-100:center_x+100]
    
    # 把处理好的最终图返回出去
    return roi


# ================= 2. 主流程 =================
def main():
    folder_path = r"D:\ycz32\Desktop\111\Learn_ai\image"
    output_folder = r"D:\ycz32\Desktop\111\Learn_ai\final_output"
    
    os.makedirs(output_folder, exist_ok=True)
    
    for filename in os.listdir(folder_path):
        if filename.lower().endswith(('.jpg', '.png', '.jpeg')):
            full_path = os.path.join(folder_path, filename)
            
            # 调用核心处理函数
            processed_img = process_image(full_path)
            
            if processed_img is None:
                print(f"❌ {filename} 处理失败")
                continue
            
            # 保存到 final_output 文件夹
            save_path = os.path.join(output_folder, filename)
            success = cv2.imwrite(save_path, processed_img)
            print(f"✅ {filename} 预处理完成（转灰+旋转+缩放+裁ROI），保存结果：{success}")

# 运行
main()
```
